# Phase 2 — Azure Infrastructure Setup

This notebook walks you through setting up:
- Resource Group
- Storage Account
- Blob Containers
- Azure Function App
- Local Azure Functions project
- Azure AI Foundry Project

Follow step by step and execute commands in terminal cells.

## Step 1 — Login to Azure

```python
!az login
```

## Step 2 — Create Resource Group

```python
!az group create --name ecommerce-agent-rg --location eastus
```

## Step 3 — Create Storage Account

```python
!az storage account create --name ecommerceagentstorage --resource-group ecommerce-agent-rg --location eastus --sku Standard_LRS
```

## Step 4 — Create Blob Containers

```python
!az storage container create --name invoices --account-name ecommerceagentstorage --auth-mode login
!az storage container create --name policies --account-name ecommerceagentstorage --auth-mode login
!az storage container create --name product-data --account-name ecommerceagentstorage --auth-mode login
!az storage container create --name faq --account-name ecommerceagentstorage --auth-mode login
```

## Step 5 — Create Azure Function App

```python
!az functionapp create --resource-group ecommerce-agent-rg --consumption-plan-location eastus --runtime python --runtime-version 3.10 --functions-version 4 --name ecommerce-agent-functions --storage-account ecommerceagentstorage
```

## Step 6 — Initialize Local Functions Project

```python
!func init ecommerce-agent-functions --python
```

## Step 7 — Create First Function

```python
!cd ecommerce-agent-functions && func new --name health_check --template "HTTP trigger" --authlevel "anonymous"
```

## Step 8 — Run Function Locally

```python
!cd ecommerce-agent-functions && func start
```

## Step 9 — Deploy to Azure

```python
!cd ecommerce-agent-functions && func azure functionapp publish ecommerce-agent-functions
```

## Step 10 — Create Azure AI Foundry Project

Go to: https://ai.azure.com

Create a new project:
- Name: ecommerce-agent-project
- Resource Group: ecommerce-agent-rg
- Region: East US
