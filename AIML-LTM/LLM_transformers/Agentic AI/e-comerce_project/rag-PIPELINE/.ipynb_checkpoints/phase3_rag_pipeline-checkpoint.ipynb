# Phase 3 — Knowledge Base for RAG

In this notebook, we will:
- Create datasets (FAQ, Policies, Products)
- Build ingestion pipeline
- Generate embeddings
- Store in FAISS vector DB
- Test retrieval


## Step 1 — Create Sample Data

```python
import os, json
os.makedirs('data', exist_ok=True)

faq = [
    {"question": "Do you offer refunds?", "answer": "Yes, within 30 days."},
    {"question": "Do you ship internationally?", "answer": "Yes."},
    {"question": "Delivery time?", "answer": "3-7 days."}
]

with open('data/faq.json', 'w') as f:
    json.dump(faq, f)

with open('data/policies.txt', 'w') as f:
    f.write("Electronics returnable within 30 days.")

products = [
    {"name": "Laptop A", "price": 900},
    {"name": "Laptop B", "price": 1200}
]

with open('data/products.json', 'w') as f:
    json.dump(products, f)
```

## Step 3 — Ingestion Pipeline

```python
from langchain.text_splitter import CharacterTextSplitter
from langchain.vectorstores import FAISS
from langchain.embeddings import HuggingFaceEmbeddings

def load_data():
    docs = []
    
    import json
    
    with open('data/faq.json') as f:
        faq = json.load(f)
        for item in faq:
            docs.append(f"Q: {item['question']} A: {item['answer']}")

    with open('data/policies.txt') as f:
        docs.append(f.read())

    with open('data/products.json') as f:
        products = json.load(f)
        for p in products:
            docs.append(f"{p['name']} costs {p['price']}")

    return docs

docs = load_data()

splitter = CharacterTextSplitter(chunk_size=200, chunk_overlap=20)
texts = splitter.split_text("\n".join(docs))

embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

db = FAISS.from_texts(texts, embeddings)
db.save_local("vector_store")

print("Vector store created")
```

## Step 4 — Test Retrieval

```python
db = FAISS.load_local("vector_store", embeddings)

query = "Can I return a laptop?"
results = db.similarity_search(query, k=2)

for r in results:
    print(r.page_content)
```
